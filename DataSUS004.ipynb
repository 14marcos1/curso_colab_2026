```python
import pandas as pd

dados = pd.read_csv("mortalidade_ac_consolidado.csv")

dados
```

```python
import matplotlib.pyplot as plt

# Calcular média e desvio-padrão
resultado = dados.groupby("GRUPO")["OBITOS"].agg(["mean", "std"])

print(resultado)

# Criar gráfico
plt.bar(
    resultado.index,
    resultado["mean"],
    yerr=resultado["std"],
    capsize=8
)

plt.ylabel("Número de óbitos")
plt.xlabel("Grupo de causas")
plt.title("Média de óbitos ± desvio-padrão")

plt.show()
```

```python
from scipy.stats import ttest_rel

circulatorio = dados[dados["GRUPO"] == "Aparelho Circulatório"]["OBITOS"]
neoplasias = dados[dados["GRUPO"] == "Neoplasias"]["OBITOS"]

teste = ttest_rel(circulatorio, neoplasias)

print("t =", teste.statistic)
print("p =", teste.pvalue)
```
