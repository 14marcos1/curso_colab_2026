Crie um código Python simples para Google Colab que permita ao usuário importar uma planilha diretamente do computador por meio de upload.

O código deve:

1. Usar `files.upload()` do `google.colab` para abrir a janela de seleção do arquivo.
2. Identificar automaticamente o nome do arquivo enviado.
3. Aceitar arquivos Excel (`.xlsx` ou `.xls`) e CSV (`.csv`).
4. Abrir o arquivo com `pandas`.
5. Para CSV, tentar automaticamente os separadores vírgula (`،`) e ponto e vírgula (`;`).
6. Armazenar os dados em um DataFrame chamado `df`.
7. Mostrar:

   * o nome do arquivo importado;
   * o número de linhas e colunas;
   * os nomes das colunas;
   * as 5 primeiras linhas da planilha.
8. Exibir uma mensagem clara caso o formato do arquivo não seja reconhecido.

O código deve ser curto, comentado e adequado para uma pessoa que está utilizando Python e Google Colab pela primeira vez. Não inclua análises estatísticas nem gráficos.


```python
# Importa as bibliotecas necessárias
from google.colab import files
import pandas as pd

print("Por favor, clique no botão abaixo para escolher o arquivo (CSV ou Excel):")

# 1. Abre a janela de seleção de arquivos do computador
uploaded = files.upload()

# 2. Identifica automaticamente o nome do arquivo enviado
if len(uploaded) > 0:
  filename = list(uploaded.keys())[0]
  print(f"\nArquivo importado com sucesso: {filename}")

  # 3. Verifica o formato e abre o arquivo com o pandas
  if filename.endswith(".csv"):
    # Tenta abrir o CSV usando vírgula como separador
    try:
      df = pd.read_csv(filename, sep=",")
      # Se o arquivo foi lido como uma única coluna, o separador provável é o ponto e vírgula
      if len(df.columns) <= 1:
        df = pd.read_csv(filename, sep=";")
    except Exception:
      df = pd.read_csv(filename, sep=";")

  elif filename.endswith((".xlsx", ".xls")):
    # Abre arquivos do Excel
    df = pd.read_excel(filename)

  else:
    # Mensagem de erro caso o formato não seja suportado
    print(
        "Erro: Formato de arquivo não reconhecido. Envie um arquivo .csv, .xlsx"
        " ou .xls."
    )

  # 4. Exibe as informações da planilha, caso tenha sido carregada corretamente
  if "df" in locals():
    print(f"\n--- Informações da Planilha ---")
    print(f"Nome do arquivo: {filename}")
    print(f"Número de linhas: {df.shape[0]}")
    print(f"Número de colunas: {df.shape[1]}")
    print(f"\nNomes das colunas:")
    print(list(df.columns))

    print(f"\nPrimeiras 5 linhas:")
    display(df.head())

else:
  print("Nenhum arquivo foi selecionado.")
```

#Análise estatística

```python
import matplotlib.pyplot as plt
import pandas as pd
import scipy.stats as stats
import seaborn as sns
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# 1. Carregamento do arquivo (utiliza o último arquivo carregado no Colab)
try:
  filename
except NameError:
  from google.colab import files

  uploaded = files.upload()
  filename = list(uploaded.keys())[0]

if filename.endswith(".csv"):
  try:
    df = pd.read_csv(filename, sep=",")
    if len(df.columns) <= 1:
      df = pd.read_csv(filename, sep=";")
  except:
    df = pd.read_csv(filename, sep=";")
elif filename.endswith((".xlsx", ".xls")):
  df = pd.read_excel(filename)

# 2. Mapeamento dos Grupos com base na sua lista de participantes (P01 a P24)
# Mapeia ID -> Grupo
mapeamento_grupos = {
    "P01": "Controle",
    "P02": "Controle",
    "P03": "Controle",
    "P04": "Controle",
    "P05": "Controle",
    "P06": "Controle",
    "P07": "Controle",
    "P08": "Controle",
    "P09": "Tratamento 1",
    "P10": "Tratamento 1",
    "P11": "Tratamento 1",
    "P12": "Tratamento 1",
    "P13": "Tratamento 1",
    "P14": "Tratamento 1",
    "P15": "Tratamento 1",
    "P16": "Tratamento 1",
    "P17": "Tratamento 2",
    "P18": "Tratamento 2",
    "P19": "Tratamento 2",
    "P20": "Tratamento 2",
    "P21": "Tratamento 2",
    "P22": "Tratamento 2",
    "P23": "Tratamento 2",
    "P24": "Tratamento 2",
}

# Identifica colunas automaticamente
col_id = next(
    (c for c in df.columns if "id" in c.lower() or "participante" in c.lower()),
    df.columns[0],
)
# Atribui a coluna de grupo com base no dicionário (ou busca coluna de grupo se já existir na planilha)
if not any("grupo" in c.lower() for c in df.columns):
  df["Grupo_Def"] = df[col_id].map(mapeamento_grupos)
else:
  col_grupo = next(c for c in df.columns if "grupo" in c.lower())
  df["Grupo_Def"] = df[col_grupo]

# Lista das análises bioquímicas a testar
variaveis = [
    "Glicose (mg/dL)",
    "Colesterol total (mg/dL)",
    "ALT (U/L)",
    "Creatinina (mg/dL)",
    "PCR (mg/L)",
]

# Tenta encontrar correspondências caso os nomes na planilha tenham pequenas variações
colunas_analise = []
for var in variaveis:
  match = next(
      (
          c
          for c in df.columns
          if var.split(" ")[0].lower() in c.lower()
          and c != "Grupo_Def"
          and c != col_id
      ),
      None,
  )
  if match:
    colunas_analise.append(match)

print(f"Variáveis encontradas na planilha para análise: {colunas_analise}\n")
print("=" * 60)

# 3. Execução dos Testes Estatísticos e Geração de Gráficos
for var in colunas_analise:
  print(f"\n--- Análise Estatística: {var} ---")

  # Remove nulos para a variável atual
  dados_analise = df[["Grupo_Def", var]].dropna()

  grupos_unicos = dados_analise["Grupo_Def"].unique()
  amostras = [
      dados_analise[dados_analise["Grupo_Def"] == g][var].values
      for g in grupos_unicos
  ]

  if len(amostras) < 2:
    print(
        f"Dados insuficientes para comparar grupos na variável {var}."
    )
    continue

  # Teste de Normalidade (Shapiro-Wilk) para cada grupo
  normal = True
  for i, g in enumerate(grupos_unicos):
    if len(amostras[i]) >= 3:
      _, p_val_shapiro = stats.shapiro(amostras[i])
      if p_val_shapiro < 0.05:
        normal = False

  # Escolha e execução do teste estatístico
  if normal and len(amostras) >= 2:
    # ANOVA de uma via
    f_val, p_val = stats.f_oneway(*amostras)
    print(f"Teste utilizado: ANOVA de Uma Via (Paramétrico)")
    print(f"Estatística F: {f_val:.4f} | p-valor: {p_val:.4f}")

    if p_val < 0.05:
      print("-> Diferença estatisticamente significativa detectada entre os grupos (p < 0.05).")
      # Teste post-hoc de Tukey
      tukey = pairwise_tukeyhsd(
          endog=dados_analise[var], groups=dados_analise["Grupo_Def"], alpha=0.05
      )
      print(tukey)
    else:
      print("-> Não houve diferença estatisticamente significativa entre os grupos (p >= 0.05).")
  else:
    # Kruskal-Wallis (Não paramétrico)
    h_val, p_val = stats.kruskal(*amostras)
    print(f"Teste utilizado: Kruskal-Wallis (Não Paramétrico)")
    print(f"Estatística H: {h_val:.4f} | p-valor: {p_val:.4f}")

    if p_val < 0.05:
      print("-> Diferença estatisticamente significativa detectada entre os grupos (p < 0.05).")
    else:
      print("-> Não houve diferença estatisticamente significativa entre os grupos (p >= 0.05).")

  # Geração do Boxplot para visualização
  plt.figure(figsize=(7, 4))
  sns.boxplot(
      x="Grupo_Def",
      y=var,
      data=dados_analise,
      palette="Set2",
      hue="Grupo_Def",
      legend=False,
  )
  sns.stripplot(
      x="Grupo_Def",
      y=var,
      data=dados_analise,
      color="black",
      size=6,
      jitter=0.2,
      alpha=0.7,
  )
  plt.title(f"Distribuição de {var} por Grupo", fontsize=11, fontweight="bold")
  plt.xlabel("Grupos", fontsize=10)
  plt.ylabel(var, fontsize=10)
  plt.tight_layout()
  plt.show()
  print("-" * 60)
```
